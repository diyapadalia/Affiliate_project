```python
import pandas as pd

conversion = pd.read_csv("Conversion.csv")
clicks = pd.read_csv("Clicks.csv")
influencer = pd.read_csv("Influencer.csv")
users = pd.read_csv("Users.csv")

result = conversion.merge(clicks, on="user_id")
result = result.merge(users[["user_id", "user_name"]], on="user_id")
result = result.merge(influencer[["influencer_id", "influencer_name"]], on="influencer_id")

result["influencer_total_revenue"] = result.groupby(
    "influencer_id"
)["subscription_revenue"].transform("sum")

result = result[[
    "influencer_name",
    "user_name",
    "subscription_plan",
    "subscription_revenue",
    "influencer_total_revenue"
]]

display(result)
```

```python
import pandas as pd

conversion = pd.read_csv("Conversion.csv")
clicks = pd.read_csv("Clicks.csv")
influencer = pd.read_csv("Influencer.csv")

# Connect conversion with influencer
data = conversion.merge(clicks, on="user_id")

# Group by influencer
result = data.groupby("influencer_id").agg(
    total_users=("user_id", "nunique"),
    total_revenue=("subscription_revenue", "sum")
).reset_index()

# Find most common plan
plan = data.groupby("influencer_id")["subscription_plan"].agg(
    lambda x: x.mode()[0]
).reset_index(name="most_common_plan")

# Add plan to result
result = result.merge(plan, on="influencer_id")

# Add influencer names
result = influencer[["influencer_id", "influencer_name"]].merge(
    result, on="influencer_id", how="left"
)

# Fill empty values
result["total_users"] = result["total_users"].fillna(0)
result["total_revenue"] = result["total_revenue"].fillna(0)
result["most_common_plan"] = result["most_common_plan"].fillna("No users")

display(result)
```

```python
import pandas as pd

conversion = pd.read_csv("Conversion.csv")
clicks = pd.read_csv("Clicks.csv")
influencer = pd.read_csv("Influencer.csv")

# Connect conversion with influencer
data = conversion.merge(clicks, on="user_id")

# Calculate revenue for each influencer
result = data.groupby("influencer_id").agg(
    total_users=("user_id", "nunique"),
    influencer_total_revenue=("subscription_revenue", "sum")
).reset_index()

# Add most common subscription plan
plan = data.groupby("influencer_id")["subscription_plan"].agg(
    lambda x: x.mode()[0]
).reset_index()

result = result.merge(plan, on="influencer_id")

# Add influencer names
result = influencer.merge(
    result, on="influencer_id", how="left"
)

# Fill missing values
result["total_users"] = result["total_users"].fillna(0)
result["influencer_total_revenue"] = result["influencer_total_revenue"].fillna(0)
result["subscription_plan"] = result["subscription_plan"].fillna("No users")

# Display influencer-wise result
result = result[[
    "influencer_name",
    "total_users",
    "subscription_plan",
    "influencer_total_revenue"
]]

display(result)

# Total revenue of all influencers
total_influencer_revenue = result["influencer_total_revenue"].sum()

print("Total Influencer Revenue:", total_influencer_revenue)
```
